```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
'''Date : 10/11/2026'''



'Let ranges be defined for B/L'


gradB = np.linspace(1e-5, 1e-11, 1000)

'''Note that phi for iron is roughly 1'''


a = np.linspace(1e-8,1e-3,1000)

rho = 7900


mu = 1.5*1e-3
mu_0 = 4*np.pi *10**-7


d1 = 1e-3
d2 = 1e-2
```

```python
def omega(mu_0, rho, gradB): 
    return np.sqrt(6/(mu_0 * rho))* gradB 


```

```python
omega_min= omega(mu_0, rho, gradB)
time_min = 2*np.pi / omega_min



omega_max_1 = 17*omega_min

time_max_1 = 2*np.pi/omega_max_1 
```

```python
plt.loglog(gradB, time_min, color = 'r' ,label = 'Timescale for minimum Omega')
plt.loglog(gradB, time_max_1, color = 'b')
plt.xlabel('GradB')
plt.ylabel('Timescale')
plt.grid(True, which = 'both', alpha = 0.5)
plt.title('Timescale for minimum Omega over GradB')
```
