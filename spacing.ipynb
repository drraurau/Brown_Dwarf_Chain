```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.special import ellipj 
#from ipywidgets import interact 
```

```python
#Global variables
#vacuum permeability 
mu_0 = 4*np.pi * 1e-7

#magnetic permeability of iron 
mu = 6.3 * 1e-3 

#magnetic field 
B = 0.1

#grain size 
a = 1e-6

#mass of grain 
M = 9.27 * 1e-26


#Length and time
L = 0.1 
t = 1
#speed of grain 
v_0 = L/t


```

```python
#magnetic moment of iron grain
def m_0():
    global mu_0 
    global mu 
    global B 
    global a
    return ((4*np.pi * a**3)/mu_0) * ((mu - mu_0)/(mu + 2*mu_0))*B


m_0 = m_0()
```

```python
#alpha parameter
def alpha(): 
    global M 
    global v_0
    global B 
    return (m_0*B)/(M*v_0**2)

def gamma(): 
    global mu_0 
    global M 
    global v_0
    global L
    return (3*m_0**2 * mu_0)/(4*np.pi * M * (L**3)*v_0**2)


al = alpha() 
gam = gamma()  

print(alpha(), gamma())
```

```python
freq = np.linspace(20,300,1000)

def r(): 
    return (-(gam* np.pi**2)/(((2*np.pi*freq)**2)- 2*al))**1/5


list = [] 


list.append(r())

print(list)
```

```python
plt.plot(freq,r())
plt.grid()
```
