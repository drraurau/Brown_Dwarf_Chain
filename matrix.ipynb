```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
'''Variables defined in the equation with a (particle radius) having a range'''

mu = 1.5*1e-3
a = np.linspace(1e-8,1e-6,100)
mu_0 = 4*np.pi *10**-7
```

```python
'''Function defined for gap distance'''

def gap_distance(radius, magmom, vacperm):
    return ((3/2)*(radius**3)*((magmom-vacperm)/(magmom + 2*vacperm)))**(1/5)



d_array = gap_distance(a, mu, mu_0)

d_min = gap_distance(a[0], mu, mu_0)
d_max = gap_distance(a[-1], mu, mu_0)
```

```python
plt.loglog(a, d_array)
plt.xlabel('Grain Radius (m)')
plt.hlines(y = (d_min, d_max), xmin = 0, xmax = 1e-6, colors = 'r', linestyle = ':')
plt.ylabel('Gap Distance (m)')
plt.grid(True, which="both", ls="-")
plt.title('Grain Radius vs Gap Distance')
```

```python

print('The minimum gap distance is '  + str(round(d_min,7)) + ' metres.')
print('The maximum gap distance is ' + str(round(d_max, 7)) + ' metres.')
```

```python
'''Function of wavelength is just 4 x d'''


wavelength = 4* d_array
max_wavelength = 4* d_max 
min_wavelength = 4 * d_min
```

```python
plt.loglog(a, wavelength)
plt.xlabel('Grain Radius (m)')
plt.hlines(y = (min_wavelength, max_wavelength), xmin = 0, xmax = 1e-6, colors = 'r', linestyle = ':')
plt.ylabel('Wavelength (m)')
plt.grid(True, which="both", ls="-")
plt.title('Grain Radius vs Wavelength')
```

```python
print('The minimum wavelength is '  + str(round(min_wavelength,7)) + ' metres.')
print('The maximum wavelength is ' + str(round(max_wavelength, 7)) + ' metres.')
```
