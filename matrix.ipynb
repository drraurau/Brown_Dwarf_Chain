```python
#Contour plot: 

#the values that are used for the equilibrium spacing 
mu = 1.5*1e-3
a = np.linspace(1e-8,1e-6,1000)
mu_0 = 4*np.pi *10**-7
L = np.linspace(1e-9,1e-6,1000)

a = np.array(a)
L = np.array(L)


[X,Y] = np.meshgrid(a,L)
fig,ax = plt.subplots(1,1)

Z = ((((3*Y**3)/(2*X**3))*((mu - mu_0)/(mu + 2*mu_0)))**(1/5))



print(X.shape, Y.shape, Z.shape)
ax.contourf(X,Y,Z)

ax.set_title('Contour Plot')
ax.set_xlabel('Grain Size')
ax.set_ylabel('Xrange')

```
