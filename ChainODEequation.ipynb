```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.special import ellipj 

import os, sys

#from ipywidgets import interact 
```

```python
def params(alpha, keff, r_0): 
    return {'alpha':alpha, 'keff_constant':keff, 'equilibrium_distance': r_0}


params = params(0.5,1,0.1)


```

```python
k = np.linspace(0,50,1000)


def omega_squared(params, k): 
    alpha = params['alpha']
    keff = params['keff_constant']
    r_0 = params['equilibrium_distance']

#defining equation 
    return np.sqrt(2*alpha + 4*keff*np.sin(k*r_0)**2)

def maximum(params): 
    r_0 = params['equilibrium_distance']

    return np.pi/r_0


a = maximum(params)/2

print(a)

```

```python
plt.plot(k,omega_squared(params,k))
plt.axvline(x = a, color = 'r', label ='maximum_k')
plt.axvline(x = maximum(params), color = 'b', label ='maximum_k')
plt.xlabel('k'); plt.ylabel(f'\omega')
```
