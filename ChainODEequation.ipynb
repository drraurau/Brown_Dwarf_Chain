```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.special import ellipj 
#from ipywidgets import interact 
```

```python
def params(alpha, keff, r_0): 
    return {'alpha':alpha, 'keff_constant':keff, 'equilibrium_distance': r_0}


params = params(0.5,4.0,1.0)


```

```python
k = np.linspace(0,50,500)


def omega_squared(params, k): 
    alpha = params['alpha']
    keff = params['keff_constant']
    r_0 = params['equilibrium_distance']

#defining equation 
    return 2*alpha - 4*keff*np.sin(k*r_0)**2


```

```python
plt.plot(k,omega_squared(params,k))
plt.xlabel('k'); plt.ylabel('omega_squared')
```
